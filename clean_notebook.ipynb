```python
import nbformat

INPUT = "4_10_model_tft.ipynb"
OUTPUT = "4_10_model_tft_clean.ipynb"

nb = nbformat.read(INPUT, as_version=4)

# Si existe metadata.widgets, la eliminamos
if "widgets" in nb.get("metadata", {}):
    del nb["metadata"]["widgets"]

nbformat.write(nb, OUTPUT)
print("Notebook limpio guardado como:", OUTPUT)
```
