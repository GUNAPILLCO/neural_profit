```python
import nbformat
# Ruta de la notebook a limpiar (input y output iguales)
INPUT = "neural_profit/6_tft_model/6_1_baseline_model.ipynb"
OUTPUT = INPUT  # sobrescribe el archivo original

# Leer notebook
nb = nbformat.read(INPUT, as_version=4)

# Eliminar metadata de widgets si existe
if "widgets" in nb.get("metadata", {}):
    del nb["metadata"]["widgets"]

# Guardar sobre la misma ruta
nbformat.write(nb, OUTPUT)

print("Notebook limpiada y sobrescrita en:", OUTPUT)
```
