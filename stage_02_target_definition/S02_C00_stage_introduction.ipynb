
# **Introducción**


Una de las decisiones más importantes en cualquier proyecto de Machine Learning aplicado a trading es la definición del **target**. Aunque gran parte de la literatura se concentra en el desarrollo de modelos predictivos cada vez más sofisticados, la calidad del target suele tener un impacto mucho mayor sobre los resultados finales.

Siguiendo el enfoque propuesto en *Machine Learning for Algorithmic Trading*, el objetivo de esta etapa no es predecir retornos exactos ni movimientos futuros del precio con precisión absoluta, sino construir y evaluar objetivos predictivos que sean consistentes, robustos y útiles desde una perspectiva operativa.

Para ello se define una arquitectura jerárquica de targets que evoluciona desde problemas simples de clasificación direccional hasta objetivos directamente relacionados con la ejecución real de operaciones.


# **Targets a investigar**

## T1 — Dirección del movimiento


Representa la formulación más simple del problema predictivo.

* 1 → el precio sube en el horizonte futuro.
* 0 → el precio no sube.

El objetivo es determinar si existe información predictiva básica sobre la dirección futura del mercado.

## T2 — Dirección con umbral

Introduce un filtro mínimo de movimiento para reducir ruido.

* 1 → movimiento positivo significativo.
* -1 → movimiento negativo significativo.
* 0 → movimiento no significativo.

Este target busca eliminar variaciones pequeñas que carecen de valor económico para una estrategia de trading.

## T3 — Outcome de trade

El foco deja de estar en el mercado y pasa a estar en la calidad de una operación.

* 1 → trade exitoso.
* 0 → trade no exitoso.

De esta forma, el modelo aprende a identificar oportunidades con potencial operativo en lugar de simplemente predecir la dirección del precio.

## T4 — Target basado en eventos



La etiqueta se construye a partir del resultado de un evento futuro.

* 1 → se alcanza primero el Take Profit.
* -1 → se alcanza primero el Stop Loss.
* 0 → ninguna barrera es alcanzada dentro del horizonte analizado.

Este enfoque se encuentra directamente alineado con la lógica de ejecución utilizada en trading sistemático.

## T5 — Probabilidades de decisión



Más que un target independiente, T5 representa una capa adicional aplicada sobre cualquiera de los targets anteriores.

El modelo estima probabilidades asociadas a cada clase y las decisiones se toman únicamente cuando la confianza supera determinados umbrales.

Esto permite filtrar señales débiles y concentrar la operativa en escenarios de mayor certidumbre.
